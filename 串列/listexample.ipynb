# 簡介

串列是個可儲存多個元素的容器

其中元素用逗點隔開，中括號包起來。

```python
scores=[78,29,56,80,99]
print(scores)
```

## 索引值

可用來存取串列中元素

注意：由0開始


```python
print(scores[0])
print(scores[1])
```

# 串列內容個數

len(a) 取得a串列中元素個數

```python
print(len(scores))
print(scores[len(scores)-1])
print(scores[-1])
```

## 新增元素
使`append()`新增至串列尾端

```python
print(scores)
scores.append(78)
scores.append(93)
print(scores)
```

```python
print(scores)
scores[2]=70
print(scores)
```

```python
weight=[12]*6
print(weight)
```

## 走訪串列的三種寫法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
for e in scores:
    print(e)
```

```python
for i,e in enumerate(scores):
    print(i,e)
```

## 計算費伯納西數列的第91個元素

1,1,2,3,5,8,...

```python
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
fib[2]=fib[1]+fib[2]
print(fib)
```

```python
i=2
fib[i]=fib[i-2]+fib[i-1]
print(fib)
for i in range(2,90):
    fib[i]=fib[i-2]+fib[i-1]

print(fib[90])
```
